```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from upsetplot import UpSet

file = "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_Allcomps.txt"

columns_all    = ["chr","start","end","A549","HeLa","HepG2","IMR90","K562","GM12878","HUVEC","HCT116"]
cell_lines_use = ["A549","HeLa","HepG2","IMR90","K562","HUVEC","HCT116"]

df = pd.read_csv(file, sep="\t", header=None, names=columns_all, dtype={"chr":"string"})
df[cell_lines_use] = df[cell_lines_use].apply(pd.to_numeric, errors="coerce").fillna(0.0)

bool_states = df[cell_lines_use].gt(0)

counts   = bool_states.groupby(cell_lines_use, sort=False).size()
percents = counts / counts.sum() * 100.0

plt.figure(figsize=(14, 7))
up = UpSet(
    percents,
    show_counts=False,
    sort_by="cardinality",
    sort_categories_by=None,
    orientation="horizontal",
    element_size=24
)
up.plot()

matrix_ax = None

ax_dict = getattr(up, "ax_dict", None)
if isinstance(ax_dict, dict) and "matrix" in ax_dict:
    matrix_ax = ax_dict["matrix"]

if matrix_ax is None:
    axes_attr = getattr(up, "axes", None)
    if isinstance(axes_attr, dict) and "matrix" in axes_attr:
        matrix_ax = axes_attr["matrix"]

if matrix_ax is None:
    for ax in plt.gcf().axes:
        ytxt = [t.get_text() for t in ax.get_yticklabels()]
        if set(cell_lines_use).issubset(set(ytxt)) and len(ytxt) >= len(cell_lines_use):
            matrix_ax = ax
            break

if matrix_ax is None:
    raise RuntimeError("Could not locate the UpSet matrix axis. Consider updating upsetplot.")

for coll in list(matrix_ax.collections):
    coll.remove()
for ln in list(matrix_ax.lines):
    ln.remove()

ytlabs = [t.get_text() for t in matrix_ax.get_yticklabels()]
ypos_by_name = {name: y for y, name in enumerate(ytlabs)}
ypos_in_order = np.array([ypos_by_name[n] for n in cell_lines_use])

order = percents.sort_values(ascending=False).index.tolist()

for x, key in enumerate(order):
    states = np.array(key, dtype=bool)
    yA = ypos_in_order[states]
    yB = ypos_in_order[~states]

    if len(yA):
        matrix_ax.scatter([x]*len(yA), yA, s=46, color="red", zorder=3)
        if len(yA) >= 2:
            matrix_ax.plot([x, x], [yA.min(), yA.max()], color="red", linewidth=1.8, zorder=2)

    if len(yB):
        matrix_ax.scatter([x]*len(yB), yB, s=34, facecolors="none", edgecolors="black", linewidth=1.2, zorder=1)

plt.suptitle("UpSet: A = Red (connected), B = Black — 7 cell lines (GM12878 excluded)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from upsetplot import UpSet

file = "genomewide_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_Allcomps.txt"

columns_all    = ["chr","start","end","A549","HeLa","HepG2","IMR90","K562","GM12878","HUVEC","HCT116"]
cell_lines_use = ["A549","HeLa","HepG2","IMR90","K562","HUVEC","HCT116"]

df = pd.read_csv(file, sep="\t", header=None, names=columns_all, dtype={"chr":"string"})
df[cell_lines_use] = df[cell_lines_use].apply(pd.to_numeric, errors="coerce").fillna(0.0)

bool_states = df[cell_lines_use].gt(0)

counts   = bool_states.groupby(cell_lines_use, sort=False).size()
percents = counts / counts.sum() * 100.0

plt.figure(figsize=(14, 7))
up = UpSet(
    percents,
    show_counts=False,
    sort_by="cardinality",
    sort_categories_by=None,
    orientation="horizontal",
    element_size=24
)
up.plot()

def get_axis(obj, key):
    d = getattr(obj, "ax_dict", None)
    if isinstance(d, dict) and key in d:
        return d[key]
    d2 = getattr(obj, "axes", None)
    if isinstance(d2, dict) and key in d2:
        return d2[key]
    return None

matrix_ax = get_axis(up, "matrix")
inter_ax  = get_axis(up, "intersections")

if matrix_ax is None:
    for ax in plt.gcf().axes:
        ytxt = [t.get_text() for t in ax.get_yticklabels()]
        if set(cell_lines_use).issubset(set(ytxt)) and len(ytxt) >= len(cell_lines_use):
            matrix_ax = ax
            break
if matrix_ax is None:
    raise RuntimeError("Could not locate the UpSet matrix axis.")

if inter_ax is None:
    for ax in plt.gcf().axes:
        yl = (ax.get_ylabel() or "").lower()
        if "intersection" in yl:
            inter_ax = ax
            break
    if inter_ax is None:
        for ax in plt.gcf().axes:
            if len(getattr(ax, "containers", [])) > 0:
                inter_ax = ax
                break
if inter_ax is None:
    raise RuntimeError("Could not locate the UpSet intersections (bars) axis.")

for coll in list(matrix_ax.collections):
    coll.remove()
for ln in list(matrix_ax.lines):
    ln.remove()

ytlabs = [t.get_text() for t in matrix_ax.get_yticklabels()]
ypos_by_name = {name: y for y, name in enumerate(ytlabs)}
ypos_in_order = np.array([ypos_by_name[n] for n in cell_lines_use])

order = percents.sort_values(ascending=False).index.tolist()

for x, key in enumerate(order):
    states = np.array(key, dtype=bool)
    yA = ypos_in_order[states]
    yB = ypos_in_order[~states]

    if len(yA):
        matrix_ax.scatter([x]*len(yA), yA, s=46, color="red", zorder=3)
        if len(yA) >= 2:
            matrix_ax.plot([x, x], [yA.min(), yA.max()], color="red", linewidth=1.8, zorder=2)

    if len(yB):
        matrix_ax.scatter([x]*len(yB), yB, s=34, color="black", zorder=2)
        if len(yB) >= 2:
            matrix_ax.plot([x, x], [yB.min(), yB.max()], color="black", linewidth=1.4, zorder=1)

all_true_key = tuple([True]*len(cell_lines_use))
if all_true_key in order:
    idx = order.index(all_true_key)

    recolored = False
    if getattr(inter_ax, "containers", None):
        for cont in inter_ax.containers:
            bars = list(cont)
            if len(bars) >= len(order):
                bars[idx].set_facecolor("red")
                bars[idx].set_edgecolor("red")
                recolored = True
                break
    if not recolored:
        rects = [p for p in inter_ax.patches if hasattr(p, "get_width")]
        if rects:
            rects_sorted = sorted(rects, key=lambda r: r.get_x())
            if idx < len(rects_sorted):
                rects_sorted[idx].set_facecolor("red")
                rects_sorted[idx].set_edgecolor("red")

plt.suptitle("UpSet: A = Red (connected), B = Black (connected) — 7 cell lines (GM12878 excluded) - GW", fontsize=14)
plt.tight_layout()
plt.savefig("CRUSH_7lines_Upset_GenomeWide.svg", format="svg", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from upsetplot import UpSet

file = "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_Allcomps.txt"

columns_all    = ["chr","start","end","A549","HeLa","HepG2","IMR90","K562","GM12878","HUVEC","HCT116"]
cell_lines_use = ["A549","HeLa","HepG2","IMR90","K562","HUVEC","HCT116"]

df = pd.read_csv(file, sep="\t", header=None, names=columns_all, dtype={"chr":"string"})
df[cell_lines_use] = df[cell_lines_use].apply(pd.to_numeric, errors="coerce").fillna(0.0)

bool_states = df[cell_lines_use].gt(0)

counts   = bool_states.groupby(cell_lines_use, sort=False).size()
percents = counts / counts.sum() * 100.0

plt.figure(figsize=(14, 7))
up = UpSet(
    percents,
    show_counts=False,
    sort_by="cardinality",
    sort_categories_by=None,
    orientation="horizontal",
    element_size=24
)
up.plot()

def get_axis(obj, key):
    d = getattr(obj, "ax_dict", None)
    if isinstance(d, dict) and key in d:
        return d[key]
    d2 = getattr(obj, "axes", None)
    if isinstance(d2, dict) and key in d2:
        return d2[key]
    return None

matrix_ax = get_axis(up, "matrix")
inter_ax  = get_axis(up, "intersections")

if matrix_ax is None:
    for ax in plt.gcf().axes:
        ytxt = [t.get_text() for t in ax.get_yticklabels()]
        if set(cell_lines_use).issubset(set(ytxt)) and len(ytxt) >= len(cell_lines_use):
            matrix_ax = ax
            break
if matrix_ax is None:
    raise RuntimeError("Could not locate the UpSet matrix axis.")

if inter_ax is None:
    for ax in plt.gcf().axes:
        yl = (ax.get_ylabel() or "").lower()
        if "intersection" in yl:
            inter_ax = ax
            break
    if inter_ax is None:
        for ax in plt.gcf().axes:
            if len(getattr(ax, "containers", [])) > 0:
                inter_ax = ax
                break
if inter_ax is None:
    raise RuntimeError("Could not locate the UpSet intersections (bars) axis.")

for coll in list(matrix_ax.collections):
    coll.remove()
for ln in list(matrix_ax.lines):
    ln.remove()

ytlabs = [t.get_text() for t in matrix_ax.get_yticklabels()]
ypos_by_name = {name: y for y, name in enumerate(ytlabs)}
ypos_in_order = np.array([ypos_by_name[n] for n in cell_lines_use])

order = percents.sort_values(ascending=False).index.tolist()

for x, key in enumerate(order):
    states = np.array(key, dtype=bool)
    yA = ypos_in_order[states]
    yB = ypos_in_order[~states]

    if len(yA):
        matrix_ax.scatter([x]*len(yA), yA, s=46, color="red", zorder=3)
        if len(yA) >= 2:
            matrix_ax.plot([x, x], [yA.min(), yA.max()], color="red", linewidth=1.8, zorder=2)

    if len(yB):
        matrix_ax.scatter([x]*len(yB), yB, s=34, color="black", zorder=2)
        if len(yB) >= 2:
            matrix_ax.plot([x, x], [yB.min(), yB.max()], color="black", linewidth=1.4, zorder=1)

all_true_key = tuple([True]*len(cell_lines_use))
if all_true_key in order:
    idx = order.index(all_true_key)

    recolored = False
    if getattr(inter_ax, "containers", None):
        for cont in inter_ax.containers:
            bars = list(cont)
            if len(bars) >= len(order):
                bars[idx].set_facecolor("red")
                bars[idx].set_edgecolor("red")
                recolored = True
                break
    if not recolored:
        rects = [p for p in inter_ax.patches if hasattr(p, "get_width")]
        if rects:
            rects_sorted = sorted(rects, key=lambda r: r.get_x())
            if idx < len(rects_sorted):
                rects_sorted[idx].set_facecolor("red")
                rects_sorted[idx].set_edgecolor("red")

plt.suptitle("UpSet: A = Red (connected), B = Black (connected) — 7 cell lines (GM12878 excluded)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgb, to_hex
from matplotlib.patches import Patch
from pathlib import Path

in_file = "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_Allcomps.txt"
out_dir = Path("Upset")
out_dir.mkdir(parents=True, exist_ok=True)
fig_path = out_dir / "consistency_pie_7lines_split_by_mag.png"
tsv_path = out_dir / "consistency_pie_7lines_split_by_mag.tsv"

all_cols = ["chr","start","end","A549","HeLa","HepG2","IMR90","K562","GM12878","HUVEC","HCT116"]
cells7   = ["A549","HeLa","HepG2","IMR90","K562","HUVEC","HCT116"]

THRESH = 50.0
MAG_METRIC = "max_abs"
labels = ["All Same","Different in 1","Different in 2","Different in 3"]
bins   = [-0.5, 0.5, 1.5, 2.5, 3.5]

dtype_map = {c: "float32" for c in cells7}
df = pd.read_csv(
    in_file, sep="\t", header=None, names=all_cols,
    usecols=cells7, dtype=dtype_map, low_memory=False
)

A = df.gt(0).to_numpy(np.int8)
nA = A.sum(axis=1)
nB = A.shape[1] - nA
majority_is_A = nA >= nB
diff_counts = np.where(majority_is_A, nB, nA)

vals = df.to_numpy(dtype=np.float32)
abs_vals = np.abs(vals)
if MAG_METRIC == "mean_abs":
    mag = abs_vals.mean(axis=1)
else:
    mag = abs_vals.max(axis=1)

mag_class = np.where(mag <= THRESH, "≤50", ">50")

cats = pd.cut(diff_counts, bins=bins, labels=labels)
xtab = pd.crosstab(cats, mag_class).reindex(index=labels, fill_value=0)
xtab = xtab.reindex(columns=["≤50", ">50"], fill_value=0)

inner_sizes = xtab.sum(axis=1).to_numpy()
outer_sizes = xtab.to_numpy().ravel(order="C")

out_df = xtab.copy()
row_tot = out_df.sum(axis=1)
out_df["total"] = row_tot
out_df["pct_≤50_in_cat"] = (out_df["≤50"] / row_tot * 100).round(2).replace([np.inf, np.nan], 0)
out_df["pct_>50_in_cat"] = (out_df[">50"] / row_tot * 100).round(2).replace([np.inf, np.nan], 0)
out_df.index.name = "category"
out_df.to_csv(tsv_path, sep="\t")

def blend(color, other=(1,1,1), frac=0.3):
    c = np.array(to_rgb(color))
    o = np.array(other)
    return to_hex((1-frac)*c + frac*o)

base_colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
outer_colors = []
explode = []
for i, base in enumerate(base_colors):
    outer_colors.append(blend(base, other=(1,1,1), frac=0.35))
    outer_colors.append(blend(base, other=(0,0,0), frac=0.25))
    explode += [0.02, 0.06]

fig, ax = plt.subplots(figsize=(8.5, 8.5))
ax.axis("equal")
start = 90

wedges_inner, _ = ax.pie(
    inner_sizes, radius=1.0, startangle=start,
    labels=labels, labeldistance=0.65,
    colors=base_colors,
    wedgeprops=dict(width=0.32, edgecolor="white")
)

wedges_outer, _ = ax.pie(
    outer_sizes, radius=1.0, startangle=start,
    labels=None,
    colors=outer_colors, explode=explode,
    wedgeprops=dict(width=0.32, edgecolor="white")
)

plt.title(f"Consistency across 7 lines (GM12878 excluded)\n"
          f"Outer ring: magnitude split by |score| ≤ {THRESH:.0f} vs > {THRESH:.0f} "
          f"({MAG_METRIC.replace('_',' ')})")

legend_patches = [
    Patch(facecolor=blend("#444", other=(1,1,1), frac=0.35), label=f"≤ {THRESH:.0f} (lighter)"),
    Patch(facecolor=blend("#444", other=(0,0,0), frac=0.25), label=f"> {THRESH:.0f} (darker)"),
]
plt.legend(handles=legend_patches, loc="upper right", frameon=False)

plt.tight_layout()
plt.savefig(fig_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Saved figure: {fig_path}")
print(f"Saved table : {tsv_path}")
```

```python
import pandas as pd
import numpy as np
import os
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, calinski_harabasz_score
import matplotlib.pyplot as plt

input_dir = "abs10"
output_dir = input_dir

os.makedirs(output_dir, exist_ok=True)

cell_line_cols = list(range(3, 13))

k_range = range(2, 16)

for file in os.listdir(input_dir):
    if not file.endswith(".bed"):
        continue

    input_file = os.path.join(input_dir, file)
    file_base = file.replace(".bed", "")

    try:
        df = pd.read_csv(input_file, sep="\t", header=None)
    except Exception as e:
        print(f"Skipping {file} (error reading file): {e}")
        continue

    if df.shape[0] <= 10:
        print(f"Skipping {file} (only {df.shape[0]} rows)")
        continue

    print(f"\nProcessing {file} with {df.shape[0]} rows...")

    comp_matrix = df.iloc[:, cell_line_cols].values

    kmeans = KMeans(n_clusters=10, random_state=42, n_init='auto')
    df["cluster"] = kmeans.fit_predict(comp_matrix)

    clustered_file = os.path.join(output_dir, f"{file_base}_kmeans10.bed")
    df.to_csv(clustered_file, sep="\t", header=False, index=False)
    print(f"Clustered file saved → {clustered_file}")

    silhouette_scores = []
    inertias = []
    ch_scores = []

    for k in k_range:
        kmeans_k = KMeans(n_clusters=k, random_state=42, n_init='auto')
        labels = kmeans_k.fit_predict(comp_matrix)

        silhouette_scores.append(silhouette_score(comp_matrix, labels))
        inertias.append(kmeans_k.inertia_)
        ch_scores.append(calinski_harabasz_score(comp_matrix, labels))

    def save_plot(scores, title, ylabel, color, suffix):
        plt.figure(figsize=(6, 5))
        plt.plot(k_range, scores, marker='o', color=color)
        plt.title(f"{title} for {file_base}")
        plt.xlabel("Number of Clusters (k)")
        plt.ylabel(ylabel)
        plt.grid(True)
        plt.tight_layout()
        out_plot = os.path.join(output_dir, f"{file_base}_{suffix}.png")
        plt.savefig(out_plot, dpi=150)
        plt.close()
        print(f"Saved plot → {out_plot}")

    save_plot(silhouette_scores, "Silhouette Score", "Silhouette Score", "blue", "silhouette")
    save_plot(inertias, "Elbow (Inertia)", "Inertia (WCSS)", "green", "elbow")
    save_plot(ch_scores, "Calinski-Harabasz Score", "CH Score", "purple", "ch_score")
```

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.multitest import multipletests
import matplotlib.pyplot as plt
import seaborn as sns
import os

comp_file = "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_Allcomps.txt"
histone_files = {
    "RNAPol2":   "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_pol2.txt",
    "H3K27ac":   "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k27ac.txt",
    "H3K4me1":   "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k4me1.txt",
    "H3K4me3":   "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k4me3.txt",
    "H3K36me3":  "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k36me3.txt",
    "H3K27me3":  "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k27me3.txt",
    "H3K9me3":   "c1_8lines_A549_HeLa_HepG2_IMR90_K562_GM12878_HUVEC_HCT116_1kb_k9me3.txt"
}
output_dir = "global_histone_regression_Pvals"
os.makedirs(output_dir, exist_ok=True)

cell_order = ["A549", "HeLa", "HepG2", "IMR90", "K562", "GM12878", "HUVEC", "HCT116"]
coord_cols = ["chr", "start", "end"]
all_cols = coord_cols + cell_order

def significance_label(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    else: return 'ns'

def read_no_header(path, sep="\t"):
    """
    Read a headerless file with columns:
    chr, start, end, A549, HeLa, HepG2, IMR90, K562, GM12878, HUVEC, HCT116
    """
    df = pd.read_csv(path, sep=sep, header=None, names=all_cols, dtype={0:str, 1:int, 2:int})
    for c in cell_order:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    return df

def melt_by_coords(df):
    return df.melt(id_vars=coord_cols, value_vars=cell_order,
                   var_name="cell", value_name="value")

comp_df = read_no_header(comp_file, sep="\t")
comp_long = melt_by_coords(comp_df).rename(columns={"value": "comp"})

rows = []
for mark, f in histone_files.items():
    hdf = read_no_header(f, sep="\t")
    hist_long = melt_by_coords(hdf).rename(columns={"value": "hist"})
    merged = comp_long.merge(hist_long, on=["chr","start","end","cell"], how="inner").dropna(subset=["comp","hist"])

    if merged.empty:
        print(f"[WARN] No overlapping rows for {mark}. Skipping.")
        continue

    X = sm.add_constant(merged["hist"].values)
    y = merged["comp"].values
    model = sm.OLS(y, X).fit()
    slope = float(model.params[1])
    r2    = float(model.rsquared)
    pval  = float(model.pvalues[1])

    print(f"{mark}: n={len(merged):,} | slope={slope:.4f} | R²={r2:.4f} | p={pval:.3e}")
    rows.append((mark, len(merged), slope, r2, pval))

summary_df = pd.DataFrame(rows, columns=["Histone", "N", "Slope", "R_squared", "P_value"])
if summary_df.empty:
    raise SystemExit("No results computed. Check input files/columns.")

_, p_adj, _, _ = multipletests(summary_df["P_value"].values, alpha=0.05, method="fdr_bh")
summary_df["P_value_BH"] = p_adj
summary_df["Signif"] = summary_df["P_value_BH"].map(significance_label)

summary_csv = os.path.join(output_dir, "global_regression_summary_with_significance.csv")
summary_df.to_csv(summary_csv, index=False)
print(f"Saved summary: {summary_csv}")

plt.figure(figsize=(7.8, 4.2))
colors = sns.color_palette("viridis", n_colors=len(summary_df))
bars = plt.bar(summary_df["Histone"], summary_df["Slope"], color=colors)
for bar, star in zip(bars, summary_df["Signif"]):
    plt.text(bar.get_x()+bar.get_width()/2, bar.get_height(),
             star, ha="center", va="bottom", fontsize=10, color="black")
plt.ylabel("Slope (histone → compartment)")
plt.title("Global Linear Regression Slopes (per mark)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
out1 = os.path.join(output_dir, "global_slopes_with_significance.png")
plt.savefig(out1, dpi=200)
plt.show()

plt.figure(figsize=(7.8, 4.2))
colors = sns.color_palette("magma", n_colors=len(summary_df))
bars = plt.bar(summary_df["Histone"], summary_df["R_squared"], color=colors)
for bar, star in zip(bars, summary_df["Signif"]):
    plt.text(bar.get_x()+bar.get_width()/2, bar.get_height(),
             star, ha="center", va="bottom", fontsize=10, color="black")
plt.ylabel("R² (variance explained)")
plt.title("Global R² per Histone Mark (BH-adjusted significance)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
out2 = os.path.join(output_dir, "global_r2_with_significance.png")
plt.savefig(out2, dpi=200)
plt.show()
```

```python
import pandas as pd
import statsmodels.api as sm
import os
import matplotlib.pyplot as plt
import seaborn as sns

base_path = "clustering"
comp_file = f"{base_path}/formatted_comps.txt"
histone_files = {
    "H3K27ac": f"{base_path}/k27ac_c1_formatted.txt",
    "H3K4me1": f"{base_path}/k4me1_c1_formatted.txt",
    "H3K4me3": f"{base_path}/k4me3_c1_formatted.txt",
    "H3K36me3": f"{base_path}/k36me3_c1_formatted.txt",
    "H3K27me3": f"{base_path}/k27me3_c1_formatted.txt",
    "H3K9me3": f"{base_path}/k9me3_c1_formatted.txt"
}
output_dir = f"{base_path}/per_cell_and_multivariate_regression"
os.makedirs(output_dir, exist_ok=True)

comp_df = pd.read_csv(comp_file, sep="\t")
cell_lines = [col for col in comp_df.columns if col != "ID"]

histone_dfs = {}
for mark, path in histone_files.items():
    df = pd.read_csv(path, sep="\t")
    histone_dfs[mark] = df

def significance_label(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    else: return 'ns'

univariate_results = []

for cell in cell_lines:
    for mark, df in histone_dfs.items():
        if cell in df.columns:
            comp = comp_df[cell]
            hist = df[cell]
            X = sm.add_constant(hist)
            y = comp
            model = sm.OLS(y, X).fit()
            slope = model.params[1]
            r2 = model.rsquared
            p_val = model.pvalues[1]
            univariate_results.append((cell, mark, slope, r2, p_val))

uni_df = pd.DataFrame(univariate_results, columns=["CellLine", "Histone", "Slope", "R_squared", "P_value"])
uni_df.to_csv(os.path.join(output_dir, "per_cell_univariate_regression.csv"), index=False)

multivariate_results = []

for cell in cell_lines:
    X_list = []
    mark_names = []
    for mark, df in histone_dfs.items():
        if cell in df.columns:
            X_list.append(df[cell])
            mark_names.append(mark)
    if len(X_list) != len(histone_files):
        continue
    X = pd.concat(X_list, axis=1)
    X.columns = mark_names
    X = sm.add_constant(X)
    y = comp_df[cell]
    model = sm.OLS(y, X).fit()

    for mark in mark_names:
        slope = model.params[mark]
        p_val = model.pvalues[mark]
        multivariate_results.append((cell, mark, slope, p_val))

    multivariate_results.append((cell, "R_squared", model.rsquared, None))

multi_df = pd.DataFrame(multivariate_results, columns=["CellLine", "Histone", "Slope_or_R2", "P_value"])
multi_df.to_csv(os.path.join(output_dir, "per_cell_multivariate_regression.csv"), index=False)

print("\n✅ Per-cell line and multivariate regression complete!")
print(f"📄 Results saved to: {output_dir}")
```

```python
import pandas as pd
import statsmodels.api as sm
import os
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

comp_file = "comp_with_clusters_k10.bed"
atac_file = "ATAC_3lines_k10clusterd.bed"
bruseq_file = "BruSeq_3lines_k10clusterd.bed"
output_dir = "global_regression_histone_atac_bru_final_3lines"
os.makedirs(output_dir, exist_ok=True)

histone_files = {
    "H3K27ac": "group_Histones_k27ac_k10clustered.txt",
    "H3K4me1": "group_Histones_k4me1_k10clustered.txt",
    "H3K4me3": "group_Histones_k4me3_k10clustered.txt",
    "H3K36me3": "group_Histones_k36me3_k10clustered.txt",
    "H3K27me3": "group_Histones_k27me3_k10clustered.txt",
    "H3K9me3": "group_Histones_k9me3_k10clustered.txt"
}

comp_df = pd.read_csv(comp_file, sep="\t", header=None)
comp_matrix = comp_df.iloc[:, [9, 10, 11]]
comp_flat = comp_matrix.values.flatten()

def load_and_zscore(path, label):
    df = pd.read_csv(path, sep="\t", header=None, usecols=[0, 1, 2])
    scaler = StandardScaler()
    z_vals = scaler.fit_transform(df)
    z_df = pd.DataFrame(z_vals, columns=["HepG2", "IMR90", "K562"])
    z_df.to_csv(os.path.join(output_dir, f"Zscored_{label}.tsv"), sep="\t", index=False)
    return z_df.values.flatten().reshape(-1, 1)

atac_z = load_and_zscore(atac_file, "ATAC")
bru_z = load_and_zscore(bruseq_file, "BruSeq")
atac_bru_combined = pd.DataFrame((atac_z + bru_z) / 2, columns=["atac_bru_combined"])

summary = []

for mark, hist_file in histone_files.items():
    hist_df = pd.read_csv(hist_file, sep="\t", header=None)
    hist_matrix = hist_df.iloc[:, [2, 3, 4]]
    hist_flat = hist_matrix.values.flatten()

    X = pd.DataFrame({
        "histone": hist_flat,
        "atac_bru": atac_bru_combined.values.flatten()
    })
    X = sm.add_constant(X)
    y = comp_flat

    model = sm.OLS(y, X).fit()

    slope_hist = model.params["histone"]
    slope_ab = model.params["atac_bru"]
    p_hist = model.pvalues["histone"]
    p_ab = model.pvalues["atac_bru"]
    r2 = model.rsquared

    summary.append((mark, slope_hist, p_hist, slope_ab, p_ab, r2))

extra_features = {
    "ATAC": atac_z.flatten(),
    "BruSeq": bru_z.flatten()
}

for label, values in extra_features.items():
    X = sm.add_constant(values)
    y = comp_flat
    model = sm.OLS(y, X).fit()

    slope = model.params[1]
    p_val = model.pvalues[1]
    r2 = model.rsquared

    summary.append((label, slope, p_val, float('nan'), float('nan'), r2))

summary_df = pd.DataFrame(summary, columns=[
    "Feature", "Slope", "P_value", "ATAC_Bru_Slope", "ATAC_Bru_P", "R_squared"
])
summary_df.to_csv(os.path.join(output_dir, "regression_summary_with_ATAC_BRU.csv"), index=False)

plt.figure(figsize=(10, 5))
bars = plt.bar(summary_df["Feature"], summary_df["Slope"], color='steelblue')

for bar, pval in zip(bars, summary_df["P_value"]):
    height = bar.get_height()
    if pd.notnull(pval):
        if pval < 0.001:
            star = '***'
        elif pval < 0.01:
            star = '**'
        elif pval < 0.05:
            star = '*'
        else:
            star = ''
        if star:
            plt.text(bar.get_x() + bar.get_width()/2, height + 0.002,
                     star, ha='center', va='bottom', fontsize=12)

plt.ylabel("Slope (β coefficient)")
plt.title("Effect of Histones, ATAC, and BruSeq on Compartment")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "slope_significance_all_features.png"))
plt.show()

plt.figure(figsize=(10, 4))
plt.bar(summary_df["Feature"], summary_df["R_squared"], color='darkorange')
plt.ylabel("R² (Variance Explained)")
plt.title("Variance Explained by Histones, ATAC, and BruSeq")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "r2_all_features.png"))
plt.show()
```

```python
1) Counts of genes in A/wA/wB/B

What it shows: How many genes live in strongly A, weak A, weak B, or strongly B compartments.

Why it matters:

Most expressed genes should be in A.

If a large chunk of genes are in B, those are probably silent/repressed.

It gives a quick overview of compartmentalization strength.

2) TSS vs Gene body bin concordance

What it shows: A heatmap of whether promoters and gene bodies agree (A–A, B–B) or disagree (A–B, B–A).

Why it matters:

Discordant genes (TSS in A, body in B or vice versa) may be in regulatory conflict — e.g., active promoters but repressive gene bodies.

This directly answers your “5kb TSS A but body B?” question.

3) Compartment value distributions (ECDF/density)

What it shows: The overall range of compartment scores for promoters or bodies.

Why it matters:

You see how polarized compartments are.

Some cell lines might show sharper separation (more extreme A/B values), others weaker.

4) Scatter: compartment vs Pol2

What it shows: Relationship between compartment strength and transcription.

Why it matters:

Strongly A → more Pol2 (should hold true if compartments reflect activity).

If slope is weak, maybe histone marks are better predictors than compartments.

5) Hexbin: TSS comp vs Pol2 at TSS

What it shows: A smoothed 2D density of promoter compartment score vs Pol2 at the TSS.

Why it matters:

Captures promoter-level regulation.

Helps see whether promoter compartment is the real driver of Pol2 loading (vs gene body compartment).

6) Correlation matrix of histone marks

What it shows: Which histone marks co-vary across genes.

Why it matters:

Active marks (K4me3, K27ac, Pol2) should positively correlate.

Repressive marks (K27me3, K9me3) should anticorrelate with active marks.

Confirms QC and reveals unexpected relationships.

7) Violin/box by promoter–body switch categories

What it shows: Pol2 (or other marks) stratified by whether TSS/body are in A or B.

Why it matters:

Genes with promoter A/body B may still have Pol2 initiation but weak elongation.

Genes with promoter B/body A are rarer but could hint at poised promoters with elongating polymerase.

8) Mark vs mark scatter (colored by compartment)

What it shows: Relationship between two histone marks (e.g., Pol2 vs K27ac) split by A/B category.

Why it matters:

In A, Pol2 and K27ac should correlate strongly.

In B, they should both be low.

Can detect sub-classes (e.g., enhancer-rich weak A vs heterochromatic B).

9) Rank-ordered metaplot

What it shows: Genes sorted by one signal (e.g., Pol2).

Why it matters:

Gives a “global” view of transcriptional activity distribution.

Lets you see whether the majority of genes are silent with only a small highly active tail.

10) PCA of histone marks

What it shows: Separation of genes based on all 7 histone marks simultaneously.

Why it matters:

PC1 usually separates active (A) from repressive (B).

PC2 may capture enhancer-poised vs constitutively active genes.

Compartment bins should cluster distinctly — a sanity check that compartments map to chromatin states.

11) TSS vs body compartment scatter

What it shows: Direct scatter of TSS vs body scores.

Why it matters:

If most points lie along the diagonal → promoter/body agree.

Off-diagonal → potential switching genes.

Lets you quantify “conflicted compartment states.”

12) Fraction of genes in each switch category

What it shows: Barplot of how frequent each TSS/body combination is (A–A, B–B, A–B, B–A).

Why it matters:

A–A should dominate (active across promoter and body).

B–B should be silent genes.

A–B fraction is interesting: active promoters but repressed gene bodies → possible stalled/poised transcription.

Summary of biological insights

Counts / distributions → How strong and polarized compartment assignments are.

TSS vs Body concordance → Identify promoter–gene body conflicts.

Histone/Pol2 vs compartment → Test whether A compartments truly reflect active transcription.

Correlations & PCA → Global chromatin state structure.

Switch analysis (A–B, B–A) → Special cases revealing genes caught between active vs repressive states.

👉 Together, these analyses let you move from “are compartments real?” to “what compartments mean for transcription and chromatin state at promoters and gene bodies.”
```

```python
import os, glob, zipfile, datetime
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

TSV_PATH = "genes_annotated_classified.tsv"

def find_tsv():
    cand = []
    for pat in [
        "./**/*classified*.tsv",
        "./**/*.tsv",
        "*classified*.tsv",
        "*.tsv",
    ]:
        cand.extend(glob.glob(pat, recursive=True))
    cand = sorted(cand, key=lambda p: (("classified" not in os.path.basename(p).lower()), p))
    return cand[0] if cand else None

if TSV_PATH is None:
    TSV_PATH = find_tsv()

if TSV_PATH is None or not os.path.exists(TSV_PATH):
    raise SystemExit(
        "Could not find a TSV. Set TSV_PATH to your genes_annotated_classified.tsv and re-run."
    )

print(f"[info] Using TSV: {TSV_PATH}")
G = pd.read_csv(TSV_PATH, sep="\t", low_memory=False)

ORDER4 = ["A","wA","wB","B"]

need = [
    "GeneBodyBin","TSSpm2kbBin","Discordant4","Discordant2",
    "Pol2_GeneBodyMean","Pol2_TSSpm2kbMean","K36me3_GeneBodyMean",
    "Pol2PauseRatio","Elongating","Paused","Bin2_Body","Bin2_TSS"
]
missing = [c for c in need if c not in G.columns]
if missing:
    print("[warn] Missing columns:", missing, "\nSome plots will be skipped but the cell will continue.")

def safe(s):
    return s.replace([np.inf, -np.inf], np.nan)

stamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
outdir = Path(f"./classified_plots_{stamp}")
outdir.mkdir(parents=True, exist_ok=True)
print(f"[info] Saving figures to: {outdir.resolve()}")

title_prefix = Path(TSV_PATH).stem


if "GeneBodyBin" in G.columns:
    fig = plt.figure(figsize=(5,3))
    cnt = G["GeneBodyBin"].value_counts().reindex(ORDER4).fillna(0).astype(int)
    plt.bar(range(len(ORDER4)), cnt.values)
    plt.xticks(range(len(ORDER4)), ORDER4)
    plt.ylabel("# genes")
    plt.title(f"{title_prefix}: GeneBody bins")
    fig.tight_layout()
    fig.savefig(outdir/"counts_bins_gene_body.png", dpi=180)
    plt.show()

if "TSSpm2kbBin" in G.columns:
    fig = plt.figure(figsize=(5,3))
    cnt = G["TSSpm2kbBin"].value_counts().reindex(ORDER4).fillna(0).astype(int)
    plt.bar(range(len(ORDER4)), cnt.values)
    plt.xticks(range(len(ORDER4)), ORDER4)
    plt.ylabel("# genes")
    plt.title(f"{title_prefix}: TSS +/- 5 kb bins")
    fig.tight_layout()
    fig.savefig(outdir/"counts_bins_tss.png", dpi=180)
    plt.show()

if {"TSSpm2kbBin","GeneBodyBin"}.issubset(G.columns):
    tbl = pd.crosstab(G["TSSpm2kbBin"], G["GeneBodyBin"]).reindex(index=ORDER4, columns=ORDER4).fillna(0)
    fig = plt.figure(figsize=(5,4.2))
    ax = plt.gca()
    im = ax.imshow(tbl.values, aspect="equal")
    ax.set_xticks(range(4)); ax.set_yticks(range(4))
    ax.set_xticklabels(ORDER4); ax.set_yticklabels(ORDER4)
    ax.set_xlabel("GeneBody bin"); ax.set_ylabel("TSS +/- 5 kb bin")
    ax.set_title(f"{title_prefix}: promoter-body concordance")
    vmax = tbl.values.max() if tbl.size else 0
    for i in range(4):
        for j in range(4):
            v = int(tbl.values[i,j]) if tbl.size else 0
            ax.text(j, i, v, ha="center", va="center")
    fig.colorbar(im, ax=ax, label="# genes")
    fig.tight_layout()
    fig.savefig(outdir/"heatmap_promoter_body.png", dpi=180)
    plt.show()

if "Paused" in G.columns:
    fig = plt.figure(figsize=(5,3))
    cnt = G["Paused"].map({True:"Paused", False:"Not"}).value_counts()
    plt.bar(range(len(cnt.index)), cnt.values)
    plt.xticks(range(len(cnt.index)), cnt.index)
    plt.ylabel("# genes")
    plt.title(f"{title_prefix}: Paused")
    fig.tight_layout()
    fig.savefig(outdir/"counts_paused.png", dpi=180)
    plt.show()

if "Elongating" in G.columns:
    fig = plt.figure(figsize=(5,3))
    cnt = G["Elongating"].map({True:"Elongating", False:"Not"}).value_counts()
    plt.bar(range(len(cnt.index)), cnt.values)
    plt.xticks(range(len(cnt.index)), cnt.index)
    plt.ylabel("# genes")
    plt.title(f"{title_prefix}: Elongating")
    fig.tight_layout()
    fig.savefig(outdir/"counts_elongating.png", dpi=180)
    plt.show()

if {"GeneBodyBin","Pol2PauseRatio"}.issubset(G.columns):
    fig = plt.figure(figsize=(5.5,3.5))
    data = [safe(G.loc[G["GeneBodyBin"]==b, "Pol2PauseRatio"].astype(float)).dropna().values for b in ORDER4]
    plt.boxplot(data, labels=ORDER4, showfliers=False)
    plt.ylabel("Pol2 TSS / Pol2 body")
    plt.title(f"{title_prefix}: pausing ratio by GeneBody bin")
    fig.tight_layout()
    fig.savefig(outdir/"box_pausing_by_bin.png", dpi=180)
    plt.show()

def ecdf_plot(values, label):
    x = np.asarray(values, dtype=float)
    x = x[~np.isnan(x)]
    if x.size == 0: 
        return None
    x = np.sort(x)
    y = np.linspace(0,1,len(x))
    plt.plot(x, y, label=label)

if {"Elongating","Paused","Pol2_GeneBodyMean"}.issubset(G.columns):
    fig = plt.figure(figsize=(6,3.2))
    mask_e = G["Elongating"]==True
    mask_p = G["Paused"]==True
    mask_o = (~mask_e) & (~mask_p)
    ecdf_plot(safe(G.loc[mask_e,"Pol2_GeneBodyMean"].astype(float)).values, "Elongating")
    ecdf_plot(safe(G.loc[mask_p,"Pol2_GeneBodyMean"].astype(float)).values, "Paused")
    ecdf_plot(safe(G.loc[mask_o,"Pol2_GeneBodyMean"].astype(float)).values, "Other")
    plt.xlabel("Pol2 (gene-body mean)"); plt.ylabel("ECDF"); plt.legend()
    plt.title(f"{title_prefix}: Pol2 distributions")
    fig.tight_layout()
    fig.savefig(outdir/"ecdf_pol2.png", dpi=180)
    plt.show()

if {"Elongating","Paused","K36me3_GeneBodyMean"}.issubset(G.columns):
    fig = plt.figure(figsize=(6,3.2))
    mask_e = G["Elongating"]==True
    mask_p = G["Paused"]==True
    mask_o = (~mask_e) & (~mask_p)
    ecdf_plot(safe(G.loc[mask_e,"K36me3_GeneBodyMean"].astype(float)).values, "Elongating")
    ecdf_plot(safe(G.loc[mask_p,"K36me3_GeneBodyMean"].astype(float)).values, "Paused")
    ecdf_plot(safe(G.loc[mask_o,"K36me3_GeneBodyMean"].astype(float)).values, "Other")
    plt.xlabel("K36me3 (gene-body mean)"); plt.ylabel("ECDF"); plt.legend()
    plt.title(f"{title_prefix}: K36me3 distributions")
    fig.tight_layout()
    fig.savefig(outdir/"ecdf_k36me3.png", dpi=180)
    plt.show()

if {"Pol2_TSSpm2kbMean","Pol2_GeneBodyMean","Elongating","Paused"}.issubset(G.columns):
    fig = plt.figure(figsize=(6,4.5))
    mask_e = G["Elongating"] == True
    mask_p = G["Paused"] == True
    mask_o = (~mask_e) & (~mask_p)
    plt.scatter(
        safe(G.loc[mask_e, "Pol2_TSSpm2kbMean"].astype(float)).values,
        safe(G.loc[mask_e, "Pol2_GeneBodyMean"].astype(float)).values,
        s=6, alpha=0.35, label="Elongating"
    )
    plt.scatter(
        safe(G.loc[mask_p, "Pol2_TSSpm2kbMean"].astype(float)).values,
        safe(G.loc[mask_p, "Pol2_GeneBodyMean"].astype(float)).values,
        s=6, alpha=0.35, label="Paused"
    )
    plt.scatter(
        safe(G.loc[mask_o, "Pol2_TSSpm2kbMean"].astype(float)).values,
        safe(G.loc[mask_o, "Pol2_GeneBodyMean"].astype(float)).values,
        s=6, alpha=0.25, label="Other"
    )
    plt.xlabel("Pol2 at TSS (+/- 5 kb)")
    plt.ylabel("Pol2 in gene body")
    plt.title(f"{title_prefix}: Pol2 promoter vs body")
    plt.legend()
    fig.tight_layout()
    fig.savefig(outdir/"scatter_pol2_tss_vs_body.png", dpi=180)
    plt.show()

if {"Bin2_Body","Elongating","Paused"}.issubset(G.columns):
    fig = plt.figure(figsize=(6,3.5))
    tmp = G.copy()
    tmp["State"] = np.where(tmp["Elongating"], "Elongating",
                     np.where(tmp["Paused"], "Paused", "Other"))
    tbl2 = pd.crosstab(tmp["Bin2_Body"], tmp["State"]).reindex(index=["A","B"]).fillna(0)
    denom = tbl2.sum(axis=1).replace(0, np.nan)
    frac = (tbl2.T / denom).T.fillna(0)
    bottom = np.zeros(len(frac.index))
    for state in ["Elongating","Paused","Other"]:
        if state in frac.columns:
            plt.bar(range(len(frac.index)), frac[state].values, bottom=bottom, label=state)
            bottom += frac[state].values
    plt.xticks(range(len(frac.index)), frac.index)
    plt.ylim(0,1); plt.ylabel("fraction"); plt.legend()
    plt.title(f"{title_prefix}: states within A vs B (body)")
    fig.tight_layout()
    fig.savefig(outdir/"stacked_states_in_AB.png", dpi=180)
    plt.show()

if {"K27ac_TSSpm2kbMean","TSSpm2kbBin","GeneBodyBin"}.issubset(G.columns):
    fig = plt.figure(figsize=(6,3.5))
    cls = np.where(G["TSSpm2kbBin"]==G["GeneBodyBin"], "Concordant", "Discordant")
    vals = [
        safe(G.loc[cls=="Concordant","K27ac_TSSpm2kbMean"].astype(float)).dropna().values,
        safe(G.loc[cls=="Discordant","K27ac_TSSpm2kbMean"].astype(float)).dropna().values
    ]
    plt.boxplot(vals, labels=["Concordant","Discordant"], showfliers=False)
    plt.ylabel("K27ac at TSS (+/- 5 kb)")
    plt.title(f"{title_prefix}: promoter acetylation vs concordance")
    fig.tight_layout()
    fig.savefig(outdir/"box_k27ac_tss_concordance.png", dpi=180)
    plt.show()

zip_path = outdir / "classified_plots_bundle.zip"
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for png in outdir.glob("*.png"):
        zf.write(png, arcname=png.name)

print(f"[done] Plots saved to: {outdir.resolve()}")
print(f"[bundle] ZIP: {zip_path.resolve()}")
```
